# Data Preprocessing Using Databricks

### Reading From CSV File

```python
df = spark.read.table("workspace.default.stores_sales_forecasting")
display(df)
```

### Cleaning the Product Name column

```python
from pyspark.sql.functions import regexp_replace, trim

# Clean 'Product Name' column by:
# 1. Replacing quotes, commas, hyphens and slashes with a space
# 2. Replacing multiple spaces with a single space
# 3. Removing leading/trailing spaces
df_clean = df.withColumn(
    "Product Name",
    trim(
        regexp_replace(
            regexp_replace(df["Product Name"], r'["\',\-\/]', ' '),
            r'\s+', ' '
        )
    )
)
display(df_clean)
```
